  > 公开属性

```python
# ============================================================
# 4. 函数
# ============================================================
def greet(name: str, greeting: str = "你好") -> str:
    """带默认参数的函数"""
    return f"{greeting}，{name}！"

print(greet("Alice"))
print(greet("Bob", greeting="Hello"))

# *args & **kwargs
def log_all(*args, **kwargs):
    print("args:", args)
    print("kwargs:", kwargs)

log_all(1, 2, 3, level="info", source="agent")

# lambda（匿名函数）
add = lambda a, b: a + b
print("lambda:", add(3, 5))
sorted_users = sorted([{"name": "B"}, {"name": "A"}], key=lambda u: u["name"])
print(sorted_users)  # [{'name': 'A'}, {'name': 'B'}]

# 常用高阶函数
nums = [1, 2, 3, 4]
print(list(map(lambda x: x * 2, nums)))       # [2, 4, 6, 8]
print(list(filter(lambda x: x > 2, nums)))     # [3, 4]
print(list(zip(["a", "b"], [1, 2])))           # [('a', 1), ('b', 2)]

# ============================================================
# 5. 异常处理
# ============================================================
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print(f"捕获异常: {e}")
except Exception as e:
    print(f"其他异常: {e}")
else:
    print("没有异常时执行")
finally:
    print("无论如何都执行")

# 手动抛出
def divide(a, b):
    if b == 0:
        raise ValueError("除数不能为 0")
    return a / b

# ============================================================
# 6. 类
# ============================================================
class Agent:
    """Agent 基类"""

    kind = "LLM Agent"  # 类属性（所有实例共享）

    def __init__(self, name: str, model: str = "qwen2.5:3b"):
        self.name = name      # 实例属性
        self.model = model
        self._tools = []      # 约定：_ 开头 = protected

    def add_tool(self, tool: str) -> "Agent":
        self._tools.append(tool)
        return self           # 链式调用

    # property — 像属性一样访问
    @property
    def tools(self) -> list:
        return self._tools

    def __repr__(self):       # 打印友好的 dunder 方法
        return f"Agent({self.name!r}, model={self.model!r})"

    def describe(self) -> str:
        t = ", ".join(self._tools) if self._tools else "无"
        return f"[{self.name}] 模型={self.model} 工具={t}"


# 继承
class CodingAgent(Agent):
    def __init__(self, name: str, language: str = "Python"):
        super().__init__(name, model="claude-sonnet-5")
        self.language = language

    def describe(self) -> str:       # 覆写父类方法
        return f"{super().describe()} | 语言={self.language}"


a1 = Agent("小助手").add_tool("search").add_tool("calculator")
a2 = CodingAgent("代码大师", "Python")
a2.add_tool("run_code")

print(a1)                 # Agent('小助手', model='qwen2.5:3b')
print(a1.describe())
print(a2.describe())
print("类属性:", Agent.kind)

# ============================================================
# 7. 文件 I/O
# ============================================================
# 写
with open("_test.txt", "w", encoding="utf-8") as f:
    f.write("第一行\n第二行\n")

# 读
with open("_test.txt", "r", encoding="utf-8") as f:
    content = f.read()
print("文件内容:", repr(content))

# 逐行读
with open("_test.txt", "r", encoding="utf-8") as f:
    for line in f:
        print("  >", line.rstrip("\n"))
```

# 装饰器、迭代器、生成器



## 迭代器

遍历list for i in [1,2,3] ; 但是遍历一个文件不可能把所有文件读入内存。  
可以for循环遍历的，就是迭代器的本质。  
 —— __iter__()  可以返回迭代器  
 —— __next__()  告知下一个值是什么，没有返回StopIteration  
所以 for i in range（num）， range 就是一个迭代器


```python
## 迭代器

class Countdown:
    """倒计时迭代器"""
    def __init__(self, start: int):
        self.current = start    # 记住当前位置

    def __iter__(self):
        return self             # 迭代器就是自己

    def __next__(self):
        if self.current <= 0:
            raise StopIteration # 没了！告诉 for 循环停下
        value = self.current
        self.current -= 1       # 每次往后挪一步
        return value

for num in Countdown(3):  # for 循环在内部调用 __iter__() 和 __next__()
    print(num)            # 3  2  1
```

## 生成器

手写迭代器太麻烦了，核心逻辑就是一个一个往外吐值。 使用生成器 yield 自动实现迭代器协议。  


```python
# 生成器 — yield
def countdown(n: int):
    while n > 0:
        yield n
        n -= 1

print(list(countdown(3)))  # [3, 2, 1]
```

## 装饰器

背景：  
——有一个需求出现在函数A、B、C、D、E……。：你想给多个函数加同样的逻辑，但 Python 的函数是一等公民——你不能改它们的源码（可能来自第三方库），也不想在每个函数里复制粘贴。  

思路： 
——不改变调用方式的基础上，引入包装+语法糖  

实现： 
——装饰器 = 一个接收函数、返回新函数的转换器 D（f） 

@timer          # 包装层 1：加计时  
@retry          # 包装层 2：加重试  
@log_args       # 包装层 3：加日志  
def call_llm(prompt):  
    return api.create(prompt)  

> 等价于: call_llm = timer(retry(log_args(call_llm)))  
>                   └──────────────────────────────┘  
>                   洋葱一样一层套一层，每层加一个功能  

```python

# ============================================================
# 8. 生成器 & 装饰器（看一眼即可，后面会深入）
# ============================================================


# 装饰器 — @ 语法糖
def timer(func):
    import time
    def wrapper(*args, **kwargs):
        t0 = time.time()
        result = func(*args, **kwargs)
        print(f"{func.__name__} 耗时 {time.time() - t0:.4f}s")
        return result
    return wrapper

@timer
def slow_add(a, b):
    import time; time.sleep(0.1)
    return a + b

print(slow_add(1, 2))

```
